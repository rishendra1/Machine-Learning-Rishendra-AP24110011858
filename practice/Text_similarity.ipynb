```python
sentence = "I love Python and I love Machine Learning"
tokens = sentence.lower().split()
```

```python
token_array = np.array(tokens)
vocabulary = np.unique(token_array)
vocabulary
```

```python
def bag_of_words(tokens, vocabulary):
  v = np.array([np.sum(np.array(tokens) == word) for word in vocabulary])
  return v
```

```python
vector = bag_of_words(tokens, vocabulary)
vector
```

```python
SentenceA = "machine learning uses numpy arrays"
SentenceB = "deep learning also uses numpy arrays heavily"
tokenA = SentenceA.lower().split()
vocabularyA = np.unique(tokenA)
tokenB = SentenceB.lower().split()
vocabularyB = np.unique(tokenB)
```

```python
vectorA = bag_of_words(tokenA, vocabularyA)
vectorB = bag_of_words(tokenB, vocabularyB)
```

```python
def similarity(vectorA, vectorB):
  if len(vectorA) < len(vectorB):
    n = len(vectorB) - len(vectorA)
    vectorA = np.append(vectorA, np.zeros(n))
  else:
    n = len(vectorB)
    vectorB = np.append(vectorB, np.zeros(n))
  dot = np.dot(vectorA,vectorB)
  magA = np.linalg.norm(vectorA)
  magB = np.linalg.norm(vectorB)
  cos = dot / (magA * magB)
  return cos
```

```python
a = similarity(vectorA, vectorB)
print(a)
```
