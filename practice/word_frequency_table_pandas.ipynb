```python
import numpy as np
import pandas as pd
```

```python
SentenceA = "I love Python"
SentenceB = "I love machine learning"
tokenA = SentenceA.lower().split()
tokenB = SentenceB.lower().split()
```

```python
tokenA
```

```python
tokenB
```

```python
combined_voc = np.unique(np.array(tokenA + tokenB))
sentence_A = [np.sum(np.array(tokenA) == word) for word in combined_voc]
sentence_B = [np.sum(np.array(tokenB) == word) for word in combined_voc]
```

```python
freq_table = pd.DataFrame({
    'word':combined_voc,
    'Sentence A':sentence_A,
    'Sentence B':sentence_B
})
```

```python
print(freq_table)
```

```python
a = "Python is easy to learn. Python is powerful and easy to use. I love Python because Python helps me learn programming."
tokens = a.lower().replace(".", "").split()
words = np.array(tokens)
freq_table = pd.DataFrame({
    'words': words,
    'frequency': [np.sum(words == word) for word in words]
    })
```
