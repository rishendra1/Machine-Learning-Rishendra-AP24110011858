```python
text = "HELLO! I am learning AIML!!!"
cleaned = ''.join(filter(lambda x : x.isalpha() or x.isdigit() or x == ' ', text))
cleaned
```

```python
token = cleaned.lower().split()
token
```

```python
vocabulary = sorted(set(token))
vocabulary
```

```python
word_id = {idx:word for word,idx in enumerate(vocabulary)}
word_id
```

```python
word_id['aiml']
```

```python
encoded_token = []

for word in token:
  encoded_token.append(word_id[word])
encoded_token
```

```python
print(f"Original tokens - {token}")
```

```python
print(f"Encoded tokens - {encoded_token}")
```

```python
import pandas as pd
a = pd.DataFrame(
    {
        'word' : vocabulary,
        'id' : [word_id[word] for word in vocabulary]
    }
)
```
